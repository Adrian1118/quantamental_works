```python
import pandas as pd
import yfinance as yf

from level2_exercise import get_connection, upsert_securities, fetch_and_store_prices, fetch_and_store_fundamentals
```

```python
conn = get_connection()
index_file_path = "SMI_EXT_constituents.csv"

constituents = pd.read_csv(index_file_path)
constituents["Ticker"] = list(constituents["Symbol"]+".SW")

tickers = list(constituents["Ticker"])
```

```python
for ticker in tickers:
    print(ticker)
    t = yf.Ticker(ticker)
    name = t.info["longName"]
    industry = t.info["industry"]
    sector = t.info["sector"]
    currency = t.info["currency"]
    print(f"Upserting Security {ticker} ({name} - {sector} - {industry} - {currency}")
    upsert_securities(conn, ticker, name, sector, industry, currency)
```

```python
fetch_and_store_prices(conn, tickers, period="1y")
```

```python
fetch_and_store_fundamentals(conn, tickers, period="1y")
```
